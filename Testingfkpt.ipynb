Inverse covariance matrix calculated:

```python
Veff = 8.8e9 # (Mpc/h)^3 
noofk = len(k) 
neff = 1e99 # infinite effective number density 
data = np.loadtxt("Final_GR_z1_CDM_EFTofLSS.dat")
k     = data[:,0]
P0    = data[:,1] + data[:,4]
P2    = data[:,2] + data[:,5]
P4    = data[:,3] + data[:,6]
Plin  = data[:,11]
fvec  = data[:,10]

noofk = len(k)
neff  = 1e99   # infinite effective number density

print("Effective survey volume =", Veff)
print("Effective galaxy density neff =", neff)

b = 1.0
beta = fvec / b

deltk = np.zeros(noofk)
deltk[:-1] = k[1:] - k[:-1]
deltk[-1]  = k[-1] - k[-2]

nk = 4*np.pi*k**2 * deltk * Veff / (2*np.pi)**3

Pterm = b*b * Plin

Cov00 = 2/nk * (
    (1 + 4/3*beta + 6/5*beta**2 + 4/7*beta**3 + 1/9*beta**4)*Pterm**2 +
    2/neff*(1 + 2/3*beta + 1/5*beta**2)*Pterm +
    1/neff**2
)

Cov02 = 2/nk * (
    (8/3*beta + 24/7*beta**2 + 40/21*beta**3 + 40/99*beta**4)*Pterm**2 +
    2/neff*(4/3*beta + 4/7*beta**2)*Pterm
)

Cov22 = 2/nk * (
    (5 + 220/21*beta + 90/7*beta**2 + 1700/231*beta**3 + 2075/1287*beta**4)*Pterm**2 +
    2/neff*(5 + 110/21*beta + 15/7*beta**2)*Pterm +
    5/neff**2
)

Cov44 = 2/nk * (
    (9 + 1404/77*beta + 104166/5005*beta**2 + 11772/1001*beta**3 + 6399/2431*beta**4)*Pterm**2 +
    1/neff*(18 + 1404/77*beta + 34722/5005*beta**2)*Pterm +
    9/neff**2
)

Cov04 = 2/nk * (
    (48/35*beta**2 + 96/77*beta**3 + 48/143*beta**4)*Pterm**2 +
    (16/35*beta**2)/neff * Pterm
)

Cov24 = 2/nk * (
    (48/7*beta + 816/77*beta**2 + 6960/1001*beta**3 + 240/143*beta**4)*Pterm**2 +
    (48/7*beta + 272/77*beta**2)/neff * Pterm
)


#  Write covariance to file-

cov_data = np.column_stack([k, Cov00, Cov02, Cov04, Cov22, Cov24, Cov44])
np.savetxt("multidata_cov.dat", cov_data, fmt="%.7f")

print("Covariance written to multidata_cov.dat")


#  Invert covariance for each k (3×3 block)

Cov00i = np.zeros(noofk)
Cov02i = np.zeros(noofk)
Cov04i = np.zeros(noofk)
Cov22i = np.zeros(noofk)
Cov24i = np.zeros(noofk)
Cov44i = np.zeros(noofk)

for i in range(noofk):
    C00 = Cov00[i]
    C02 = Cov02[i]
    C04 = Cov04[i]
    C22 = Cov22[i]
    C24 = Cov24[i]
    C44 = Cov44[i]

    DET = (
        C00*(C22*C44 - C24*C24)
        - C02*(C02*C44 - C24*C04)
        + C04*(C02*C24 - C22*C04)
    )

    Cov00i[i] = (C22*C44 - C24*C24) / DET
    Cov02i[i] = -(C02*C44 - C24*C04) / DET
    Cov04i[i] = (C02*C24 - C22*C04) / DET
    Cov22i[i] = (C00*C44 - C04*C04) / DET
    Cov24i[i] = -(C00*C24 - C02*C04) / DET
    Cov44i[i] = (C00*C22 - C02*C02) / DET


#  Write inverse covariance

icov_data = np.column_stack([k, Cov00i, Cov02i, Cov04i, Cov22i, Cov24i, Cov44i])
np.savetxt("multidata_invcov.dat", icov_data, fmt="%.7f")

print("Inverse covariance written to multidata_invcov.dat")

#  Compute error bars σℓ = sqrt(1 / iCovℓℓ)
sigma0 = np.sqrt(1 / Cov00i)
sigma2 = np.sqrt(1 / Cov22i)
sigma4 = np.sqrt(1 / Cov44i)

#  Write errorbars file

err_data = np.column_stack([k, P0, sigma0, P2, sigma2, P4, sigma4])
np.savetxt("multidata_errorbars.dat", err_data, fmt="%.7f")

print("Errorbars written to multidata_errorbars.dat")
```

CHI^2 calculation:


This has python script of a  for loop that makes the calculation of chi^2 much simpler to calculate and also takes less manual interactions to switch between models.

```python
model = 'FR5_CDM'  # Options: 'FR5_CDM', 'FR4_CDM', 'FR5_HALO', 'FR4_HALO'

# LOADING EXACT MULTIPOLES (FROM MGCOPTER)
if model == 'FR5_CDM':
    b1_val = b1
    b2_val = b2
    bs2_val = bs2
    b3nl_val = b3nl
    fR0 = 1e-5
    P0_data, P2_data, P4_data = fr5_cdm_exact_P0, fr5_cdm_exact_P2, fr5_cdm_exact_P4
if model == 'FR4_CDM':
    b1_val = b1
    b2_val = b2
    bs2_val = bs2
    b3nl_val = b3nl
    fR0 = 1e-4
    P0_data, P2_data, P4_data = fr4_cdm_exact_P0, fr4_cdm_exact_P2, fr4_cdm_exact_P4

    # If halo bias also needs to be calcuated, define bias values in the nuisance parameters section

if model == 'FR5_HALO':
    b1_val = b1H
    b2_val = b2H
    bs2_val = bs2H
    b3nl_val = b3nlH
    fR0 = 1e-5
    P0_data, P2_data, P4_data = fr5_halo_exact_P0, fr5_halo_exact_P2, fr5_halo_exact_P4
if model == 'FR4_HALO':
    b1_val = b1H
    b2_val = b2H
    bs2_val = bs2H
    b3nl_val = b3nlH
    fR0 = 1e-4
    P0_data, P2_data, P4_data = fr4_halo_exact_P0, fr4_halo_exact_P2, fr4_halo_exact_P4

#  RUN FKPT MODEL FOR GIVEN COUNTERTERMS
def compute_fkpt_model(a0, a2, a4):

    fkpt = run_fkpt(pk_name = 'PS_MG.txt', 
                   zout = z_pk, Om = Om, h = h, 
                   b1 = b1_val, b2 = b2_val, bs2 = bs2_val, b3nl = b3nl_val,
                   alpha0 = a0, alpha2 = a2, alpha4 = a4, 
                   ctilde = ctilde, pshotp = pshotp,  alpha0shot = alpha0shot, alpha2shot = alpha2shot,
                   model = 'HS', fR0 = fR0)

    return fkpt

# LOAD INVERSE COVARIANCE
tmp = np.loadtxt("multidata_invcov.dat")
iCov = np.zeros((len(P0_data), 3, 3))
iCov[:, 0, 0] = tmp[:, 1]
iCov[:, 0, 1] = tmp[:, 2]
iCov[:, 0, 2] = tmp[:, 3]
iCov[:, 1, 1] = tmp[:, 4]
iCov[:, 1, 2] = tmp[:, 5]
iCov[:, 2, 2] = tmp[:, 6]
# Symmetry
iCov[:, 1, 0] = iCov[:, 0, 1]
iCov[:, 2, 0] = iCov[:, 0, 2]
iCov[:, 2, 1] = iCov[:, 1, 2]


# DEFINE K BINS AND KMAX VALUES
kbins = {kk: idx for idx, kk in enumerate(k)}
kmin = 0.01
kmax_list = [
    1.044009e-01, 1.173923e-01, 1.320002e-01,
    1.484260e-01, 1.668957e-01, 1.876638e-01,
    2.110161e-01
]

#  CHI-SQUARED COMPUTATION
def compute_chi2(P0_model, P2_model, km):

    chi_list = []
    # for km in kmax_list:
    indices = [kbins[kk] for kk in k if kmin <= kk <= km]
    Nbins = len(indices)
    Ndof = 2 * Nbins - 3  # 3 EFT parameters (alpha0, alpha2, alpha4)
    chi2_total = 0.0

    for i in indices:
        resid = np.array([
            P0_data[i] - P0_model[i],
            P2_data[i] - P2_model[i]
        ])

        Cinv = np.array([
            [iCov[i, 0, 0], iCov[i, 0, 2]],
            [iCov[i, 2, 0], iCov[i, 2, 2]]
        ])

        chi2_total += resid @ Cinv @ resid

    chi_list.append(chi2_total / Ndof)

    return chi_list

alpha0_range = np.arange(-3, 3, 0.1)
alpha2_range = np.arange(8, 12, 0.5)
alpha4_range = np.arange(10, 15, 0.5)

a0 = -8                       
a2 = -6        
a4 = -14

best_hits = []

counter_0 = [
    {"start": -20, "end": 10, "step": 3}, #k0
    {"start": -20, "end": 1,  "step": 3}, #k1
    {"start": -10, "end": 3,  "step": 5}, #k2
    {"start": -10, "end": 7,  "step": 5}, #k3
    {"start": -11, "end": 8,  "step": 1}, #k4
    {"start": -12, "end": 10, "step": 1}, #k5
    {"start": -15, "end": 10, "step": 1}, #k6
]

counter_1 = [
    {"start": -30, "end": 30, "step": 5}, #k0
    {"start": -30, "end": 30, "step": 5}, #k1
    {"start": -18, "end": 3,  "step": 4}, #k2
    {"start": -30, "end": 10, "step": 1}, #k3
    {"start": -1,  "end": 3,  "step": 1}, #k4
    {"start": -50, "end": 10, "step": 5}, #k5
    {"start": -15, "end": 15, "step": 3}, #k6
]

counter_2 = [
    {"start": -30, "end": 20, "step": 5}, #k0
    {"start": -30, "end": 15, "step": 5}, #k1
    {"start": -18, "end": 10, "step": 1}, #k2
    {"start": 0,   "end": 5,  "step": 1}, #k3
    {"start": -40, "end": 29, "step": 1}, #k4
    {"start": -50, "end": 10, "step": 5}, #k5
    {"start": -40, "end": 35, "step": 1}, #k6
]

# kmax list
kmax_list = [1.044009e-01, 1.173923e-01, 1.320002e-01, 1.484260e-01, 1.668957e-01, 1.876638e-01, 2.110161e-01]

# Open file for saving readable results
outfile = open(f"eft_counterterm_safe_{model}.txt", "a")
outfile.write(f"# {model} alpha0   alpha2   alpha4   chi(k1) chi(k2) ... chi(k9)\n")

# Loop using index ki
for ki in range(6, len(kmax_list)):
    print(ki, counter_0[ki], counter_1[ki], counter_2[ki], kmax_list[ki])
    kmax = kmax_list[ki]

    c0 = counter_0[ki]
    c1 = counter_1[ki]
    c2 = counter_2[ki]

    i_start, i_end, i_step = c0["start"], c0["end"], c0["step"]
    j_start, j_end, j_step = c1["start"], c1["end"], c1["step"]
    k_start, k_end, k_step = c2["start"], c2["end"], c2["step"]

    i = i_start
    while i <= i_end:
        j = j_start
        while j <= j_end:
            _k = k_start
            while _k <= k_end:
                fkpt = compute_fkpt_model(i, j, _k)
                P0_model = fkpt["mono"]
                P2_model = fkpt["quad"]
                chi = compute_chi2(P0_model, P2_model, kmax)
                print(f"kmax={kmax:.6f}, alpha0={i:.6f}, alpha2={j:.6f}, alpha4={_k:.6f}, chi2={chi[0]:.6f}")
                if chi[0] <= 600:
                    # Save to file
                    print(f"kmax={kmax:.6f}, alpha0={i:.6f}, alpha2={j:.6f}, alpha4={_k:.6f}, chi2={chi[0]:.6f}")
                    line = f"kmax:{kmax:.6f}  {i:.6f}  {j:.6f}  {_k:.6f}  " + "  ".join(f"{c:.6f}" for c in chi) 
                    outfile.write(line + "\n")
                _k += k_step
            j += j_step
        i += i_step


for c in chi:
    best_hits.append((f"{c:.6f}"))

outfile.close()
print("\nSaved all scan results to eft_counterterm_safe.txt")
print("SUMMARY: Found", best_hits, "parameter sets")
```
